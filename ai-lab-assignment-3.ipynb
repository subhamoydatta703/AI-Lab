## Q2. Write a python program for KNN classsifier

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load dataset
iris = pd.read_csv("/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv")
```

```python
# print data shape
print(f"Data Shape: {iris.shape}")
```

```python
# print first 5 rows
print(f"First 5 rows: \n{iris.head()}")
```

```python
x=iris.iloc[:,1:4]
y= iris["species"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train, y_train)
```

```python
y_pred=knn.predict(x_test)
```

```python
print(y_pred)
```

```python
print("Actual species: ",y_test.values)
```

```python
accuracy= accuracy_score(y_test, y_pred)
print("Accuracy: ",accuracy*100,"%")
```

```python
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
print("Classification report:\n",classification_report(y_test, y_pred))
```
