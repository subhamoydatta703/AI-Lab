```python
from collections import deque
```

```python
def is_safe(row, col):
    if row < 0:
        return False
    if row >= 3:
        return False
    if col < 0:
        return False
    if col >= 3:
        return False
    return True
```

```python
def find_blank(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return i, j
```

```python
def get_neighbours(board):
    blank_row, blank_col = find_blank(board)
    neighbours = []

    for row_offset, col_offset in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
        new_row = blank_row + row_offset
        new_col = blank_col + col_offset

        if is_safe(new_row, new_col):
            new_board = [row[:] for row in board]

            
            new_board[blank_row][blank_col], new_board[new_row][new_col] = \
                new_board[new_row][new_col], new_board[blank_row][blank_col]

            neighbours.append(new_board)

    return neighbours
```

```python
def print_board(board):
    for row in board:
        print(row)
    print()


board = [
    [1, 2, 3],
    [4, 5, 6],
    [0, 7, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


queue = deque()
queue.append((board, 0))

visited = set()
visited.add(str(board))

while queue:
    current_board, steps = queue.popleft()

    print("Level =", steps)
    print_board(current_board)

    if current_board == goal:
        print("Goal Found")
        print("Minimum Steps =", steps)
        break

    for neighbour in get_neighbours(current_board):
        if str(neighbour) not in visited:
            visited.add(str(neighbour))
            queue.append((neighbour, steps + 1))
else:
    print("Goal not reachable")
```
