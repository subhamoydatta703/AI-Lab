### Q1. Show me a python program that takes user(student) inputs such as  attendence, marks, assignment performence and internal assesment and use If Then rules to classify the student as excellent, good, average or poor.

```python
def classify_student():
    print("--- Student Performance Classifier ---")

    # Getting input
    attendance = float(input("Enter attendance percentage: "))
    marks = float(input("Enter marks (out of 100): "))
    assignment = float(input("Enter assignment performance (out of 10): "))
    internal = float(input("Enter internal assessment score (out of 20): "))

    # Simple weighted logic
    total_score = (marks * 0.5) + (assignment * 2) + (internal * 1.5)

    if attendance >= 90 and total_score >= 80:
        classification = "Excellent"
    elif attendance >= 75 and total_score >= 60:
        classification = "Good"
    elif attendance >= 60 and total_score >= 40:
        classification = "Average"
    else:
        classification = "Poor"

    print(f"\nStudent is classified as: {classification}")


classify_student()
```

### Q2. Develop a Python program to implement a rule based expert system for medical diagnosis. The system should accept  symptopms such as fever, cough, headche and sore throat and use IF Then to identify the possib ble disease.

```python
def diagnose_condition():
    print("--- Symptom Checker (Not Medical Advice!) ---")

    # Getting input
    fever = input("Do you have a fever? (yes/no): ").lower() == "yes"
    cough = input("Do you have a persistent cough? (yes/no): ").lower() == "yes"
    fatigue = input("Are you feeling unusually tired? (yes/no): ").lower() == "yes"
    shortness_of_breath = input("Do you have shortness of breath? (yes/no): ").lower() == "yes"

    # Logic tree
    if shortness_of_breath:
        diagnosis = "Potential respiratory issue. Please consult a doctor immediately."
    elif fever and cough and fatigue:
        diagnosis = "Possibility of viral infection like Flu or COVID-19. Monitor and rest."
    elif cough and not fever:
        diagnosis = "Possible allergies or mild irritation. Keep an eye on it."
    elif fatigue and not fever:
        diagnosis = "Could be stress or lack of sleep. Try to rest."
    else:
        diagnosis = "Symptoms are inconclusive. If you feel unwell, see a professional."

    print(f"\nDiagnosis suggestion: {diagnosis}")


diagnose_condition()
```
