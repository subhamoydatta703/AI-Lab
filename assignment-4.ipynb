# BFS

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []
    visited.add(start)
    while queue:
        node = queue.popleft()
        traversal.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return traversal
result = bfs(graph, 'A')
print("BFS Traversal:")
print("->".join(result))
G = nx.DiGraph()
for node in graph:
    for neighbor in graph[node]:
        G.add_edge(node, neighbor)
        pos = {
            'A': (0, 3),
            'B': (-1, 2),
            'C': (1, 2),
            'D': (-1.5, 1),
            'E': (-0.5, 1),
            'F': (1, 1),
            'G': (-0.5, 0)
        }
        plt.figure(figsize=(8, 6))
        nx.draw(
            G, pos,
            with_labels=True,
            node_color='yellow',
            node_size=2000,
            font_size=14,
            font_weight='bold',
            edge_color='blue',
            arrows=True,
            width=2
        )
        plt.title("BFS Graphical Representation", fontsize=15)
        plt.axis('off')
        plt.show()
```

# DFS

```python
import networkx as nx
import matplotlib.pyplot as plt
def dfs(graph, start):
    if start not in graph:
        raise ValueError(f"Start node '{start}' not found in graph.")
    visited = set([start])
    stack = [start] 
    traversal = []
    while stack:
        node = stack.pop()
        traversal.append(node)
        for neighbor in reversed(graph[node]):
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)
    return traversal
result = dfs(graph, 'A')
print("DFS Traversal:")
print("->".join(result))
G = nx.DiGraph()
for node in graph:
    for neighbor in graph[node]:
        G.add_edge(node, neighbor)
        pos = {
            'A': (0, 3),
            'B': (-1, 2),
            'C': (1, 2),
            'D': (-1.5, 1),
            'E': (-0.5, 1),
            'F': (1, 1),
            'G': (-0.5, 0)
        }
        plt.figure(figsize=(8, 6))
        nx.draw(
            G, pos,
            with_labels=True,
            node_color='yellow',
            node_size=2000,
            font_size=14,
            font_weight='bold',
            edge_color='blue',
            arrows=True,
            width=2
        )
        plt.title("DFS Graphical Representation", fontsize=15)
        plt.axis('off')
        plt.show()
```
